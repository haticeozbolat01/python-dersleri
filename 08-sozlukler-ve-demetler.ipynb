## Sözlük oluşturmak ve değere ulaşmak

```python
ogrenci = {"ad": "Ayşe", "yas": 21, "bolum": "Bilgisayar"}

print(ogrenci["ad"])
print(ogrenci.get("sehir"))
print(ogrenci.get("sehir", "bilinmiyor"))
```

## Eklemek, değiştirmek, silmek

```python
ogrenci["sehir"] = "Adana"    # yeni anahtar: eklendi
ogrenci["yas"] = 22           # var olan anahtar: güncellendi
print(ogrenci)

del ogrenci["bolum"]
print(ogrenci)

print("ad" in ogrenci)
print("Ayşe" in ogrenci)
```

## Sözlüğü döngüyle gezmek

```python
notlar = {"Ayşe": 85, "Mert": 45, "Zeynep": 92}

for isim in notlar:
    print(isim)

for isim, puan in notlar.items():
    print(f"{isim}: {puan}")

print(list(notlar.keys()))
print(list(notlar.values()))
print(sum(notlar.values()))
```

## Kelime saymak

```python
metin = "bir elma bir armut bir muz"
sayac = {}

for kelime in metin.split():
    sayac[kelime] = sayac.get(kelime, 0) + 1

print(sayac)
```

## Demetler

```python
konum = (36.99, 35.33)       # enlem, boylam
print(konum[0])

enlem, boylam = konum
print(enlem, boylam)

print(type((5,)), type((5)))

mesafeler = {("Adana", "Mersin"): 70}
print(mesafeler[("Adana", "Mersin")])
```

Demet değiştirilemez. Aşağıdaki hücre bilerek hata veriyor:

```python
konum[0] = 1
```

## Mini program: sınıf defteri

Çalıştırınca öğrenci adı ve notu isteyecek. Bitirmek için `bitti` yaz.

```python
sinif = {}

while True:
    isim = input("Öğrenci adı (bitirmek için 'bitti'): ")
    if isim == "bitti":
        break
    puan = float(input(f"{isim} notu: "))
    sinif[isim] = puan

print()
for isim, puan in sinif.items():
    if puan >= 50:
        print(f"{isim}: {puan} (Geçti)")
    else:
        print(f"{isim}: {puan} (Kaldı)")

en_iyi = max(sinif, key=sinif.get)
print(f"Sınıf birincisi: {en_iyi} ({sinif[en_iyi]})")
print(f"Sınıf ortalaması: {sum(sinif.values()) / len(sinif):.1f}")
```

## Cevaplar

```python
# 1
meyveler = {"elma": 30, "muz": 45}
meyveler["armut"] = 25
meyveler["elma"] = 35
print(sum(meyveler.values()))

# 2
baskentler = {"Türkiye": "Ankara", "Japonya": "Tokyo", "İspanya": "Madrid"}
for ulke, sehir in baskentler.items():
    print(f"{ulke} → {sehir}")

# 3
renk = (255, 128, 0)
r, g, b = renk
print(r, g, b)

# 4
cumle = "python python veri bilim veri python"
sayac = {}
for kelime in cumle.split():
    sayac[kelime] = sayac.get(kelime, 0) + 1
print(sayac)
print(max(sayac, key=sayac.get))
```
