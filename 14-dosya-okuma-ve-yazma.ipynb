## İlk dosyamızı yazalım

`open()` fonksiyonu bir dosyayı açar. İki şey veriyoruz: dosyanın adı ve ne yapacağımız (`"w"` = yaz). `encoding="utf-8"` ise Türkçe karakterlerin (ç, ğ, ı, ö, ş, ü) doğru kaydedilmesi için.

```python
dosya = open("notlar.txt", "w", encoding="utf-8")
dosya.write("Bugün dosyalara giriş yaptım.\n")
dosya.write("Şimdi ne öğrendiğimi okuyacağım.\n")
dosya.close()
print("Dosya yazıldı.")
```

`\n` satır sonu demek. Yazmazsan iki cümle aynı satıra yapışır.

`close()` dosyayı kapatır. Kapatmayı unutursan yazdıkların dosyaya tam olarak kaydedilmeyebilir. Bunu unutmamanın en kolay yolu `with` kullanmak.

## with: kapatmayı Python'a bırak

`with` bloğu bitince dosya kendiliğinden kapanır. Bundan sonra hep bu yolu kullanacağız.

```python
with open("notlar.txt", "r", encoding="utf-8") as dosya:
    icerik = dosya.read()

print(icerik)
```

## Üç temel mod

| Mod | Anlamı | Dosya yoksa | Dosya varsa |
|---|---|---|---|
| `"r"` | oku | hata verir | baştan okur |
| `"w"` | yaz | oluşturur | **içini siler**, baştan yazar |
| `"a"` | ekle | oluşturur | sonuna ekler |

`"w"` modunun dosyanın içini sildiğini unutma. Eski notlarını korumak istiyorsan `"a"` kullan.

```python
with open("notlar.txt", "a", encoding="utf-8") as dosya:
    dosya.write("Bu satır sonradan eklendi.\n")

with open("notlar.txt", "r", encoding="utf-8") as dosya:
    print(dosya.read())
```

## Satır satır okumak

Dosyayı bir `for` döngüsüne verirsen her turda bir satır gelir. Satır sonundaki `\n` karakterini `strip()` ile atıyoruz (Ders 11'deki string metotlarını hatırla).

```python
with open("notlar.txt", "r", encoding="utf-8") as dosya:
    for numara, satir in enumerate(dosya, start=1):
        print(numara, satir.strip())
```

Tüm satırları bir listeye almak istersen `readlines()` ya da `read().splitlines()` kullanabilirsin:

```python
with open("notlar.txt", "r", encoding="utf-8") as dosya:
    satirlar = dosya.read().splitlines()

print(satirlar)
print("Satır sayısı:", len(satirlar))
```

## Sayıları yazmak

`write()` sadece metin kabul eder. Sayıyı yazmadan önce `str()` ile metne, okurken de `int()` ile geri sayıya çevirmen gerekir.

```python
puanlar = [85, 92, 78]

with open("puanlar.txt", "w", encoding="utf-8") as dosya:
    for puan in puanlar:
        dosya.write(str(puan) + "\n")

with open("puanlar.txt", "r", encoding="utf-8") as dosya:
    okunan = [int(satir) for satir in dosya]

print(okunan)
print("Ortalama:", sum(okunan) / len(okunan))
```

## Dosya yoksa ne olur?

Olmayan bir dosyayı `"r"` ile açmaya çalışırsan `FileNotFoundError` alırsın. Ders 13'te öğrendiğimiz `try` ve `except` burada işe yarıyor.

```python
try:
    with open("olmayan_dosya.txt", "r", encoding="utf-8") as dosya:
        print(dosya.read())
except FileNotFoundError:
    print("Böyle bir dosya yok. Önce oluşturman gerekiyor.")
```

## CSV dosyaları

Veri analizinde en sık göreceğin dosya türü CSV: her satır bir kayıt, değerler virgülle ayrılmış. Python'un hazır `csv` modülü bu işi kolaylaştırıyor.

```python
import csv

ogrenciler = [
    ["ad", "bolum", "not"],
    ["Ayşe", "Bilgisayar", 88],
    ["Mehmet", "Endüstri", 74],
    ["Zeynep", "Bilgisayar", 95],
]

with open("ogrenciler.csv", "w", encoding="utf-8", newline="") as dosya:
    yazici = csv.writer(dosya)
    yazici.writerows(ogrenciler)

with open("ogrenciler.csv", "r", encoding="utf-8") as dosya:
    print(dosya.read())
```

`newline=""` parametresi, bazı sistemlerde satırlar arasında boş satır oluşmasını engelliyor. CSV yazarken eklemeyi alışkanlık hâline getir.

Okurken `csv.DictReader` her satırı bir sözlüğe çeviriyor, böylece sütunlara adıyla ulaşabiliyorsun:

```python
with open("ogrenciler.csv", "r", encoding="utf-8") as dosya:
    for satir in csv.DictReader(dosya):
        print(satir["ad"], "->", satir["not"])
```

Dikkat: CSV'den okunan her şey metin olarak gelir. `satir["not"]` bir sayı değil, `"88"` gibi bir metin. Hesap yapacaksan `int()` ile çevir.

## Mini proje: unutmayan alışveriş listesi

Program her açıldığında listeyi dosyadan yüklesin, yeni ürün eklenince dosyaya kaydetsin.

```python
LISTE_DOSYASI = "alisveris.txt"

def listeyi_yukle():
    try:
        with open(LISTE_DOSYASI, "r", encoding="utf-8") as dosya:
            return dosya.read().splitlines()
    except FileNotFoundError:
        return []

def urun_ekle(urun):
    with open(LISTE_DOSYASI, "a", encoding="utf-8") as dosya:
        dosya.write(urun + "\n")

def listeyi_goster():
    liste = listeyi_yukle()
    if not liste:
        print("Liste boş.")
    for sira, urun in enumerate(liste, start=1):
        print(f"{sira}. {urun}")
```

```python
listeyi_goster()
urun_ekle("süt")
urun_ekle("ekmek")
urun_ekle("domates")
print("---")
listeyi_goster()
```

Bu hücreyi bir kez daha çalıştır: ürünler iki kez eklenecek. Çünkü liste artık programın içinde değil, dosyada duruyor. Sıfırlamak için aşağıdaki hücreyi çalıştır.

```python
with open(LISTE_DOSYASI, "w", encoding="utf-8") as dosya:
    pass
listeyi_goster()
```

`"w"` modunda açıp hiçbir şey yazmamak dosyanın içini boşaltıyor. `pass`, "burada yapılacak bir şey yok" demek.

## Alıştırmalar

1. Kendine `gunluk.txt` adında bir dosya oluştur ve içine üç satır yaz. Sonra dosyayı okuyup kaç satır olduğunu ekrana yazdır.
2. `puanlar.txt` dosyasına iki yeni puan ekle (dosyanın içini silmeden). Sonra en yüksek puanı bul.
3. `ogrenciler.csv` dosyasını `DictReader` ile oku ve sadece Bilgisayar bölümündeki öğrencilerin not ortalamasını hesapla.
4. Alışveriş listesine `urun_sil(urun)` fonksiyonu ekle. İpucu: listeyi yükle, ürünü çıkar, kalanları `"w"` ile baştan yaz.

## Cevaplar

```python
# 1
with open("gunluk.txt", "w", encoding="utf-8") as dosya:
    dosya.write("Sabah kahvaltı yaptım.\n")
    dosya.write("Öğlen Python çalıştım.\n")
    dosya.write("Akşam kitap okudum.\n")
with open("gunluk.txt", "r", encoding="utf-8") as dosya:
    print("Satır sayısı:", len(dosya.read().splitlines()))

# 2
with open("puanlar.txt", "a", encoding="utf-8") as dosya:
    dosya.write("99\n")
    dosya.write("64\n")
with open("puanlar.txt", "r", encoding="utf-8") as dosya:
    print("En yüksek puan:", max(int(s) for s in dosya))

# 3
notlar = []
with open("ogrenciler.csv", "r", encoding="utf-8") as dosya:
    for satir in csv.DictReader(dosya):
        if satir["bolum"] == "Bilgisayar":
            notlar.append(int(satir["not"]))
print("Bilgisayar ortalaması:", sum(notlar) / len(notlar))

# 4
def urun_sil(urun):
    liste = listeyi_yukle()
    if urun not in liste:
        print(urun, "listede yok.")
        return
    liste.remove(urun)
    with open(LISTE_DOSYASI, "w", encoding="utf-8") as dosya:
        for kalan in liste:
            dosya.write(kalan + "\n")

urun_ekle("elma")
urun_ekle("peynir")
urun_sil("elma")
urun_sil("muz")
listeyi_goster()
```
