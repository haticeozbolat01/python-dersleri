## Cevap: Ortalama neden 80.0 çıkıyor?

`/` her zaman `float` döndürür. `//` küsuratı atar, `%` kalanı verir.

```python
print(240 / 3)   # normal bölme, sonuç hep float
print(240 // 3)  # tam bölme, küsuratı atar
print(7 / 2)
print(7 // 2)
print(7 % 2)     # bölmeden kalan
```

## Tip dönüştürme

```python
print(int("5") + 3)     # metni sayıya çevirdik
print("5" + str(3))     # sayıyı metne çevirdik
print(float("3.5"))
print(int(3.9))         # int() yuvarlamaz, küsuratı atar
print(bool(0), bool("merhaba"))
```
