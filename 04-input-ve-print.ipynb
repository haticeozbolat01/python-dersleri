## print: ekrana yazmak

Virgülle ayırdığın değerleri `print` aralarına boşluk koyarak yazar.

```python
print("Merhaba", "dünya")
```

`sep` değerlerin arasına ne konacağını belirler.

```python
print("Merhaba", "dünya", sep="-")
print("2026", "10", "27", sep="/")
```

`end` satırın sonuna ne yazılacağını belirler. Varsayılan değeri yeni satırdır.

```python
print("Yükleniyor", end="... ")
print("tamam")
```

f-string ile değişkeni metnin içine yerleştir.

```python
isim = "Ayşe"
print(f"Merhaba {isim}, hoş geldin!")
```

## input: kullanıcıdan bilgi almak

```python
isim = input("Adın ne? ")
print("Merhaba", isim)
```

## input her zaman metin verir

Aşağıdaki hücreye bir yıl yaz: sonuç yine `str` çıkacak.

```python
dogum = input("Doğum yılın? ")
print(type(dogum))
```

Hesap yapmak için `int()` ile sayıya çevir.

```python
dogum = input("Doğum yılın? ")
yas = 2026 - int(dogum)
print(f"Bu yıl {yas} yaşına giriyorsun.")
```

## Mini program: not ortalaması hesaplayıcı

```python
isim = input("Adın: ")
not1 = float(input("1. sınav: "))   # input'un sonucunu hemen float'a çeviriyoruz
not2 = float(input("2. sınav: "))
not3 = float(input("3. sınav: "))

ortalama = (not1 + not2 + not3) / 3
print(f"{isim}, ortalaman {ortalama:.1f}")
```

## Sık karşılaşılan hatalar

Aşağıdaki hücreler bilerek hata veriyor. Hata mesajını oku, sonra bir sonraki hücreye geç.

```python
int("abc")   # ValueError: harfleri sayıya çeviremez
```

```python
float("85,5")   # ValueError: Python ondalık ayırıcı olarak nokta bekler
```

```python
print("70" + "85")            # 7085: metinler yan yana eklendi
print(int("70") + int("85"))  # 155
```

## Alıştırmalar

1. `print` ile tek satırda `Python | kolay | eğlenceli` yazdır. İpucu: üç ayrı değer ve `sep`.
2. Kullanıcıya en sevdiği diziyi sor, cevabı `En sevdiğin dizi: ...` şeklinde yazdır.
3. Kullanıcıdan Celsius cinsinden bir sıcaklık al, Fahrenheit'a çevir ve yazdır. Formül: F = C × 9 / 5 + 32.
4. Aşağıdaki program neden `7085` yazıyor? Düzelt.

```python
a = input("1. sayı: ")   # 70 yaz
b = input("2. sayı: ")   # 85 yaz
print(a + b)
```

## Cevaplar

1. `print("Python", "kolay", "eğlenceli", sep=" | ")`
2. `dizi = input("En sevdiğin dizi? ")` ve ardından `print(f"En sevdiğin dizi: {dizi}")`
3. `c = float(input("Sıcaklık (°C): "))`, `fahrenheit = c * 9 / 5 + 32`, `print(f"{c} °C = {fahrenheit} °F")`
4. `input` metin verdiği için `a + b` iki metni yan yana ekledi. Düzeltme: `print(int(a) + int(b))` sonucu `155` olur.
