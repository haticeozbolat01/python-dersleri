## 1. Tarih ve saat nesneleri

`datetime` modülünde en çok kullanacağın üç şey var: `date` (sadece tarih), `datetime` (tarih ve saat) ve `timedelta` (iki an arasındaki süre).

```python
from datetime import date, datetime, timedelta

dogum = date(2000, 3, 21)
toplanti = datetime(2026, 11, 20, 14, 30)

print(dogum)
print(toplanti)
print(dogum.year, dogum.month, dogum.day)
print(toplanti.hour, toplanti.minute)
```

```python
print(date.today())
print(datetime.now())
```

## 2. Tarih aritmetiği: timedelta

İki tarihi çıkarınca bir `timedelta` (süre) elde edersin. Bir tarihe süre ekleyince de yeni bir tarih.

```python
baslangic = date(2026, 11, 20)
yilbasi = date(2027, 1, 1)

fark = yilbasi - baslangic
print(fark)
print(fark.days, "gün")

print("Bir hafta sonra:", baslangic + timedelta(days=7))
print("100 gün önce:", baslangic - timedelta(days=100))
```

## 3. Haftanın günü

```python
GUNLER = ["Pazartesi", "Salı", "Çarşamba", "Perşembe", "Cuma", "Cumartesi", "Pazar"]

gun = date(2026, 11, 20)
print(gun.weekday())
print(GUNLER[gun.weekday()])
```

`weekday()` Pazartesi için 0, Pazar için 6 döndürüyor. Türkçe gün adı için kendi listemizi kullanıyoruz.

## 4. Tarihi metne çevirmek: strftime

`strftime` (string format time) tarihi istediğin biçimde metne çeviriyor. Her `%` ile başlayan kod bir parçayı temsil ediyor.

```python
an = datetime(2026, 11, 20, 9, 5)

print(an.strftime("%d.%m.%Y"))
print(an.strftime("%d/%m/%Y %H:%M"))
print(an.strftime("%Y-%m-%d"))
```

| Kod | Anlamı | Örnek |
|---|---|---|
| `%d` | gün | 20 |
| `%m` | ay (sayı) | 11 |
| `%Y` | yıl (4 haneli) | 2026 |
| `%H` | saat (24 saat) | 09 |
| `%M` | dakika | 05 |

Ay ve gün adlarını yazdıran `%B` ve `%A` kodları bilgisayarın dil ayarına göre İngilizce gelebilir; Türkçe ad için yukarıdaki gibi kendi listeni kullanmak en güvenlisi.

## 5. Metni tarihe çevirmek: strptime

Veri setlerinde tarihler genelde metin olarak gelir. `strptime` (string parse time) tersini yapıyor. Biçimi doğru söylemen gerekiyor.

```python
metin = "20.11.2026 14:30"
an = datetime.strptime(metin, "%d.%m.%Y %H:%M")
print(an, type(an))

print(date.fromisoformat("2026-11-20"))
```

`fromisoformat` en yaygın biçim olan `YYYY-AA-GG` için kısa yol.

Biçim tutmazsa `ValueError` alırsın. Ders 13'teki `try` ve `except` burada işe yarar:

```python
for metin in ["20.11.2026", "2026/11/20", "31.02.2026"]:
    try:
        print(datetime.strptime(metin, "%d.%m.%Y").date())
    except ValueError:
        print(f"'{metin}' okunamadı")
```

Üçüncü örnek biçim olarak doğru ama 31 Şubat diye bir gün yok; Python bunu da yakalıyor.

## 6. Mini uygulama: kaç gün kaldı?

```python
def kac_gun_kaldi(hedef_metin, bugun):
    hedef = datetime.strptime(hedef_metin, "%d.%m.%Y").date()
    kalan = (hedef - bugun).days
    if kalan > 0:
        return f"{hedef_metin} tarihine {kalan} gün var ({GUNLER[hedef.weekday()]})."
    if kalan == 0:
        return "Bugün o gün!"
    return f"O tarih {-kalan} gün önce geçti."

kontrol = date(2026, 11, 20)
print(kac_gun_kaldi("01.01.2027", kontrol))
print(kac_gun_kaldi("20.11.2026", kontrol))
print(kac_gun_kaldi("29.10.2026", kontrol))
```

Ders 18'deki gibi bugünün tarihini fonksiyona dışarıdan veriyoruz; gerçek kullanımda `date.today()` gönderebilirsin.

## Cevaplar
