## Örnek 1: Python'ı okumak kolay

Çalıştırmadan önce kodu satır satır oku ve ekrana ne yazacağını tahmin et. `#` ile başlayan satırlar açıklamadır; Python onları atlar.

```python
# Bir öğrencinin üç sınav notu
notlar = [70, 80, 90]

# Ortalamayı bul: notların toplamını, not sayısına böl
ortalama = sum(notlar) / len(notlar)

# Sonucu ekrana yazdır
print("Ortalaman:", ortalama)

# Ortalamaya göre karar ver
if ortalama >= 50:
    print("Dersi geçtin")
else:
    print("Bir daha denemelisin")
```

Sonuç neden 80 değil de **80.0**? Cevabı Python Dersi 3'te, veri tiplerinde.

## Örnek 2: Kısa yazmak zamanla gelir

Aşağıdaki iki yol da aynı listeyi üretir: her sayının karesini. Kısa yolu şimdi çözmeye çalışma; döngüleri 6. derste, liste üreteçlerini (list comprehension) 12. derste işleyeceğiz.

```python
sayilar = [1, 2, 3, 4, 5]

# Uzun yol: boş bir liste aç, döngüyle tek tek ekle
kareler = []
for sayi in sayilar:
    kareler.append(sayi * sayi)

# Kısa yol: aynı işi tek satırda yap (list comprehension)
kareler_kisa = [sayi * sayi for sayi in sayilar]

print(kareler)
print(kareler_kisa)
```

**1. Çıktıyı tahmin et.** Notlar `[40, 45, 50]` olursa program ekrana ne yazar?

```python
# Bir öğrencinin üç sınav notu
notlar = [40, 45, 50]

# Ortalamayı bul: notların toplamını, not sayısına böl
ortalama = sum(notlar) / len(notlar)

# Sonucu ekrana yazdır
print("Ortalaman:", ortalama)

# Ortalamaya göre karar ver
if ortalama >= 50:
    print("Dersi geçtin")
else:
    print("Bir daha denemelisin")
```

**2. Sınırı değiştir.** `50` yerine `60` yazarsan `[70, 80, 90]` notları için sonuç değişir mi?

```python
# Bir öğrencinin üç sınav notu
notlar = [70, 80, 90]

# Ortalamayı bul: notların toplamını, not sayısına böl
ortalama = sum(notlar) / len(notlar)

# Sonucu ekrana yazdır
print("Ortalaman:", ortalama)

# Ortalamaya göre karar ver
if ortalama >= 60:
    print("Dersi geçtin")
else:
    print("Bir daha denemelisin")
```
