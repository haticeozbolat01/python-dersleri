## Metin bir karakter dizisidir

```python
s = "Merhaba Dünya"

print(len(s))      # karakter sayısı (boşluk dahil)
print(s[0])        # ilk karakter
print(s[-1])       # son karakter
print(s[0:7])      # 0'dan 7'ye kadar (7 hariç)
print(s[::-1])     # ters çevir
```

## Türkçe karakter tuzağı: I ve İ

```python
print("istanbul".upper())   # İSTANBUL olmalıydı
print("ISPARTA".lower())    # ısparta olmalıydı
print(len("İSTANBUL".lower()), len("istanbul"))
```

```python
def tr_lower(metin):
    return metin.replace("I", "ı").replace("İ", "i").lower()

def tr_upper(metin):
    return metin.replace("i", "İ").replace("ı", "I").upper()

print(tr_lower("ISPARTA İZMİR"))
print(tr_upper("istanbul ılgaz"))
```

## Mini program: metin analizcisi

```python
def metin_analizi(metin):
    """Bir metnin kelime sayısını, en uzun kelimesini ve en sık geçen kelimesini döndürür."""
    kelimeler = [k.strip(".,!?;:").lower() for k in metin.split()]
    sayac = {}
    for k in kelimeler:
        sayac[k] = sayac.get(k, 0) + 1
    return {
        "karakter": len(metin),
        "kelime": len(kelimeler),
        "en_uzun": max(kelimeler, key=len),
        "en_sik": max(sayac, key=sayac.get),
    }

metin = "Python öğrenmek eğlenceli. Python ile veri analizi de eğlenceli!"
for ad, deger in metin_analizi(metin).items():
    print(f"{ad:<10} {deger}")
```

```python
def palindrom_mu(metin):
    """Boşlukları ve noktalama işaretlerini yok sayarak metin tersten de aynı mı diye bakar."""
    temiz = "".join(h for h in tr_lower(metin) if h.isalnum())
    return temiz == temiz[::-1]

print(palindrom_mu("Ey Edip Adanada pide ye"))
print(palindrom_mu("Python"))
```

```python
def maskele(eposta):
    kullanici, alan = eposta.split("@")
    return kullanici[0] + "*" * (len(kullanici) - 1) + "@" + alan

print(maskele("ayse.yilmaz@ornek.com"))
```

## Cevaplar

```python
# 1
ad, soyad = "ayşe", "yılmaz"
print(f"{tr_upper(soyad)}, {ad.capitalize()}")

# 2
def sesli_say(cumle):
    return sum(1 for h in tr_lower(cumle) if h in "aeıioöuü")
print(sesli_say("Merhaba Dünya"))

# 3
yil, ay, gun = "2026-11-04".split("-")
print(f"{gun}.{ay}.{yil}")

# 4
def sifre_guclu_mu(sifre):
    uzun = len(sifre) >= 8
    rakam = any(h.isdigit() for h in sifre)
    buyuk = any(h.isupper() for h in sifre)
    return uzun and rakam and buyuk
print(sifre_guclu_mu("python"), sifre_guclu_mu("Python2026"))
```
