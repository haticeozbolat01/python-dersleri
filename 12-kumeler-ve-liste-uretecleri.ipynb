## Küme: tekrarsız eleman topluluğu

```python
sehirler = {"Adana", "Mersin", "Adana", "Hatay"}
print(sehirler)
print(len(sehirler))

ziyaretler = ["Adana", "İzmir", "Adana", "Ankara", "İzmir", "Adana"]
farkli = set(ziyaretler)
print(farkli, len(farkli))
```

Kümede sıra yoktur. Aşağıdaki hücre bilerek hata veriyor:

```python
sehirler[0]
```

## Eklemek, silmek, aramak

```python
sehirler.add("Osmaniye")
sehirler.add("Adana")      # zaten var, bir şey değişmez
sehirler.discard("Hatay")
sehirler.discard("Rize")    # yok ama hata vermez
print(sehirler)
print("Mersin" in sehirler)

bos = set()                 # {} boş sözlük demek!
print(type(bos), type({}))
```

## Küme işlemleri

```python
python_sinifi = {"Ayşe", "Mert", "Zeynep", "Can"}
r_sinifi = {"Zeynep", "Can", "Elif"}

print(python_sinifi | r_sinifi)   # birleşim: en az birinde olanlar
print(python_sinifi & r_sinifi)   # kesişim: ikisinde de olanlar
print(python_sinifi - r_sinifi)   # fark: sadece Python'da olanlar
print(python_sinifi ^ r_sinifi)   # simetrik fark: sadece birinde olanlar
```

Küme yazdırılınca elemanların sırası değişebilir. Sıralı görmek için `sorted` kullan:

```python
print(sorted(python_sinifi & r_sinifi))
```

## Liste üreteci

```python
notlar = [70, 85, 40, 90, 45]

gecenler = [n for n in notlar if n >= 50]
artirilmis = [n + 5 for n in notlar]
harfler = ["Geçti" if n >= 50 else "Kaldı" for n in notlar]

print(gecenler)
print(artirilmis)
print(harfler)
```

## Sözlük ve küme üreteçleri

```python
isimler = ["Ayşe", "Mert", "Zeynep"]
uzunluklar = {isim: len(isim) for isim in isimler}
print(uzunluklar)

notlar_sozluk = {"Ayşe": 85, "Mert": 45, "Zeynep": 92}
gecenler = {isim: puan for isim, puan in notlar_sozluk.items() if puan >= 50}
print(gecenler)

epostalar = ["ayse@gmail.com", "mert@outlook.com", "can@gmail.com"]
alanlar = {e.split("@")[1] for e in epostalar}
print(sorted(alanlar))
```

## Mini program: iki listeyi karşılaştırmak

```python
def karsilastir(liste1, liste2):
    """İki listenin ortak, yalnızca birincide ve yalnızca ikincide olan elemanlarını döndürür."""
    a, b = set(liste1), set(liste2)
    return {
        "ortak": sorted(a & b),
        "sadece_1": sorted(a - b),
        "sadece_2": sorted(b - a),
    }

gecen_hafta = ["süt", "ekmek", "yumurta", "peynir", "süt"]
bu_hafta = ["ekmek", "domates", "peynir", "zeytin"]

for ad, liste in karsilastir(gecen_hafta, bu_hafta).items():
    print(f"{ad:<9} {liste}")
```

## Bonus: kümede aramak neden hızlı?

```python
import time

sayilar_liste = list(range(1_000_000))
sayilar_kume = set(sayilar_liste)

bas = time.perf_counter()
for _ in range(100):
    999_999 in sayilar_liste
liste_sure = time.perf_counter() - bas

bas = time.perf_counter()
for _ in range(100):
    999_999 in sayilar_kume
kume_sure = time.perf_counter() - bas

print(f"Liste: {liste_sure:.4f} sn, küme: {kume_sure:.6f} sn")
print(f"Küme yaklaşık {liste_sure / kume_sure:.0f} kat hızlı")
```
