## Kalıp ve kurabiye

**Sınıf** (class) bir kurabiye kalıbı gibi: Şekli bir kez tarif edersin. **Nesne** (object) o kalıptan çıkan her bir kurabiye: Hepsi aynı şekilde ama her birinin kendi rengi, kendi süslemesi olabilir.

Geçen dersteki her görev bir sözlüktü. Şimdi görevi bir sınıfa çevirelim.

```python
class Gorev:
    def __init__(self, baslik):
        self.baslik = baslik
        self.bitti = False

    def tamamla(self):
        self.bitti = True

    def __str__(self):
        isaret = "[x]" if self.bitti else "[ ]"
        return f"{isaret} {self.baslik}"


g1 = Gorev("Ders 17'yi oku")
g2 = Gorev("Alıştırmaları çöz")
g1.tamamla()

print(g1)
print(g2)
print(type(g1))
```

Satır satır:

- `class Gorev:` kalıbın adı. Sınıf adları büyük harfle başlar.
- `__init__` nesne oluşturulurken otomatik çalışan özel metot. Başlangıç değerlerini burada veriyoruz.
- `self` "bu nesnenin kendisi" demek. `self.baslik`, o nesneye ait başlık.
- `tamamla` bir **metot**: sınıfın içinde tanımlanmış fonksiyon. `g1.tamamla()` diye çağrılıyor.
- `__str__` nesneyi `print` ettiğimizde ne yazılacağını belirliyor.

## Her nesne kendi verisini taşır

```python
print(g1.bitti, g2.bitti)
g2.baslik = "Alıştırmaların hepsini çöz"
print(g2)
```

`g1`'i tamamlamak `g2`'yi etkilemedi. Aynı kalıptan çıktılar ama ayrı kurabiyeler.

## İkinci örnek: kuralı olan bir nesne

Sınıfların asıl gücü, veriyle birlikte kuralları da bir arada tutmak. Bir kumbara düşünelim: Para eklenebilir, çekilebilir ama içindekinden fazlası çekilemez.

```python
class Kumbara:
    def __init__(self, sahibi, bakiye=0):
        self.sahibi = sahibi
        self.bakiye = bakiye
        self.hareketler = []

    def yatir(self, tutar):
        if tutar <= 0:
            print("Tutar sıfırdan büyük olmalı.")
            return
        self.bakiye += tutar
        self.hareketler.append(f"+{tutar}")

    def cek(self, tutar):
        if tutar > self.bakiye:
            print(f"Yetersiz bakiye. Kumbarada {self.bakiye} TL var.")
            return
        self.bakiye -= tutar
        self.hareketler.append(f"-{tutar}")

    def __str__(self):
        return f"{self.sahibi} kumbarası: {self.bakiye} TL"


k = Kumbara("Zeynep")
k.yatir(200)
k.yatir(50)
k.cek(500)
k.cek(80)
print(k)
print(k.hareketler)
```

500 TL çekmeye çalışınca kumbara izin vermedi. Bu kural sınıfın içinde yazılı; kumbarayı kullanan kodun bunu hatırlaması gerekmiyor.

`hareketler` listesini `__init__` içinde oluşturduğumuza dikkat et. Böylece her kumbaranın kendi hareket listesi oluyor.

## Nesneleri bir listede tutmak

Nesneler de sayılar ya da metinler gibi listeye konabilir.

```python
gorevler = [Gorev("Python çalış"), Gorev("Spor yap"), Gorev("Kitap oku")]
gorevler[1].tamamla()

for g in gorevler:
    print(g)

biten = sum(g.bitti for g in gorevler)
print(f"{len(gorevler)} görevden {biten} tanesi bitti.")
```

## Alıştırmalar

1. `Kitap` adında bir sınıf yaz: `baslik`, `yazar` ve `sayfa` özellikleri olsun. `__str__` metodu "Başlık - Yazar (sayfa)" yazdırsın.
2. `Kitap` sınıfına `okunan_sayfa` özelliği ve `oku(sayfa)` metodu ekle. Okunan sayfa toplam sayfayı geçemesin.
3. `Kitap` sınıfına, okumanın yüzde kaçının bittiğini döndüren `ilerleme()` metodu ekle.

## Cevaplar

```python
class Kitap:
    def __init__(self, baslik, yazar, sayfa):
        self.baslik = baslik
        self.yazar = yazar
        self.sayfa = sayfa
        self.okunan_sayfa = 0

    def oku(self, sayfa):
        self.okunan_sayfa = min(self.okunan_sayfa + sayfa, self.sayfa)

    def ilerleme(self):
        return round(self.okunan_sayfa / self.sayfa * 100)

    def __str__(self):
        return f"{self.baslik} - {self.yazar} ({self.sayfa})"


kitap = Kitap("Örnek Roman", "Ad Soyad", 240)
print(kitap)
kitap.oku(120)
kitap.oku(150)
print(kitap.okunan_sayfa, f"%{kitap.ilerleme()}")
```
