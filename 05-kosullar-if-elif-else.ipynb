## Karşılaştırma: doğru mu, yanlış mı?

```python
print(5 > 3)
print(5 == 3)
print(10 != 7)
print(7 >= 7)
print("elma" == "elma")
```

## if: bir koşul doğruysa

```python
ortalama = 81.7

if ortalama >= 50:
    print("Geçtin!")
```

## else: değilse

`ortalama` değerini değiştirip tekrar çalıştır.

```python
ortalama = 42

if ortalama >= 50:
    print("Geçtin!")
else:
    print("Kaldın.")
```

## elif: birden fazla seçenek

```python
ortalama = 81.7

if ortalama >= 85:
    harf = "A"
elif ortalama >= 70:
    harf = "B"
elif ortalama >= 50:
    harf = "C"
else:
    harf = "F"

print(f"Harf notun: {harf}")
```

### Sıra önemli

Koşullar ters sırada olursa program hata vermez ama yanlış sonuç üretir.

```python
ortalama = 92

if ortalama >= 50:      # 92 ≥ 50 doğru, burada durur
    harf = "C"
elif ortalama >= 70:
    harf = "B"
elif ortalama >= 85:
    harf = "A"

print(harf)
```

## and, or, not

```python
sicaklik = 28
yagmur = False

if sicaklik > 25 and not yagmur:
    print("Piknik zamanı!")

print(True and False, True or False, not True)
```

## Mini program: not değerlendirici

Çalıştırınca adını ve üç notu soracak.

```python
isim = input("Adın: ")
not1 = float(input("1. sınav: "))
not2 = float(input("2. sınav: "))
not3 = float(input("3. sınav: "))
ortalama = (not1 + not2 + not3) / 3

if ortalama >= 85:
    harf = "A"
elif ortalama >= 70:
    harf = "B"
elif ortalama >= 50:
    harf = "C"
else:
    harf = "F"

print(f"{isim}, ortalaman {ortalama:.1f}, harf notun {harf}.")
if harf == "F":
    print("Bu dönem olmadı, bütünlemede görüşürüz.")
else:
    print("Tebrikler, dersi geçtin!")
```
