## Tekerleği yeniden icat etme

Karekök almak, rastgele sayı üretmek, bugünün tarihini bulmak... Bunların hepsini sıfırdan yazabilirsin ama gerek yok. Başkaları yazmış, test etmiş ve paylaşmış. Python'da bu hazır kodlara **modül** diyoruz.

Bir modülü kullanmak için `import` yazıyoruz:

```python
import math

print(math.sqrt(144))
print(math.pi)
print(math.floor(7.8), math.ceil(7.2))
```

`math.sqrt` şu demek: `math` modülünün içindeki `sqrt` fonksiyonu. Nokta, "içindeki" anlamına geliyor.

## import yazmanın üç yolu

```python
# 1. Modülün tamamını al, adıyla kullan
import random
print(random.randint(1, 6))

# 2. Sadece ihtiyacın olanı al
from math import sqrt
print(sqrt(81))

# 3. Uzun adlara kısa takma ad ver
import statistics as st
print(st.mean([70, 85, 90]))
```

Üçüncü yolu veri biliminde çok göreceksin: `import pandas as pd`, `import numpy as np`. Bunlar herkesin kullandığı kalıplaşmış kısaltmalar.

`from modul import *` diye her şeyi tek seferde almak da mümkün ama önerilmez: Hangi fonksiyonun nereden geldiğini kaybedersin.

## Python'la birlikte gelen modüller

Python'u kurduğunda yüzlerce modül de birlikte gelir. Buna **standart kütüphane** deniyor. Birkaç kullanışlı örnek:

```python
from datetime import date

bugun = date.today()
yilbasi = date(bugun.year + 1, 1, 1)
print("Yılbaşına", (yilbasi - bugun).days, "gün var.")
```

```python
from collections import Counter

kelimeler = "elma armut elma kiraz elma armut".split()
print(Counter(kelimeler).most_common(2))
```

```python
import random

random.seed(42)
takim = ["Ayşe", "Mehmet", "Zeynep", "Can"]
print("Sunumu yapacak kişi:", random.choice(takim))
random.shuffle(takim)
print("Konuşma sırası:", takim)
```

`random.seed(42)` ile her çalıştırmada aynı "rastgele" sonucu alıyoruz. Bunu sildiğinde her seferinde farklı biri seçilir.

Bir modülün içinde ne olduğunu merak edersen `dir()` ve `help()` kullanabilirsin:

```python
print([ad for ad in dir(math) if not ad.startswith("_")][:15])
help(math.gcd)
```

## Kendi modülünü yazmak

Modül dediğimiz şey aslında sıradan bir `.py` dosyası. Ders 14'te dosya yazmayı öğrenmiştik; şimdi bir Python dosyası yazıp onu modül olarak kullanalım.

```python
kod = '''
def selamla(isim):
    return f"Merhaba {isim}, Python Dersi 15'e hoş geldin!"

def ortalama(sayilar):
    return sum(sayilar) / len(sayilar)

PI_YAKLASIK = 3.14
'''

with open("araclarim.py", "w", encoding="utf-8") as dosya:
    dosya.write(kod)
print("araclarim.py oluşturuldu.")
```

```python
import araclarim

print(araclarim.selamla("Hatice"))
print(araclarim.ortalama([80, 90, 100]))
print(araclarim.PI_YAKLASIK)
```

Dosya ile `import` eden not defteri aynı klasörde olduğu için Python dosyayı buldu. Büyük projelerde kod böyle parçalara bölünüyor: her dosya bir konuyla ilgileniyor, diğerleri onu `import` ediyor.
