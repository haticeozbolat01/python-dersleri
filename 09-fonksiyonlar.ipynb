## Fonksiyonu matematikteki gibi düşün

## Parametre: fonksiyona bilgi vermek

## return: fonksiyondan sonuç almak

```python
def ortalama(notlar):
    return sum(notlar) / len(notlar)

print(ortalama(ayse))
mert_ort = ortalama(mert)
print(mert_ort + 5)
```

`print` ile `return` farkı:

```python
def yazdir_ortalama(notlar):
    print(sum(notlar) / len(notlar))

sonuc = yazdir_ortalama(mert)
print(sonuc)
```

## Varsayılan değer ve isimli argüman

```python
def gecti_mi(puan, sinir=50):
    return puan >= sinir

print(gecti_mi(55))
print(gecti_mi(55, 60))
print(gecti_mi(sinir=40, puan=45))
```

## Birden fazla sonuç döndürmek

```python
def istatistik(notlar):
    return min(notlar), max(notlar), sum(notlar) / len(notlar)

en_dusuk, en_yuksek, ort = istatistik(zeynep)
print(en_dusuk, en_yuksek, ort)
print(istatistik(mert))
```

## Docstring: fonksiyona açıklama yazmak

```python
def harf_notu(puan):
    """Puanı harf notuna çevirir: 85+ AA, 70+ BA, 50+ CC, altı FF."""
    if puan >= 85:
        return "AA"
    elif puan >= 70:
        return "BA"
    elif puan >= 50:
        return "CC"
    return "FF"

print(harf_notu(92), harf_notu(73), harf_notu(50), harf_notu(30))
help(harf_notu)
```

## Mini program: sınıf defteri, fonksiyonlarla

```python
def rapor(sinif):
    """Her öğrencinin ortalamasını, harf notunu ve durumunu yazdırır."""
    for isim, notlar in sinif.items():
        ort = ortalama(notlar)
        durum = "Geçti" if gecti_mi(ort) else "Kaldı"
        print(f"{isim:<8} {ort:5.1f}  {harf_notu(ort)}  {durum}")

sinif = {"Ayşe": ayse, "Mert": mert, "Zeynep": zeynep, "Can": [30, 55, 41]}
rapor(sinif)
```

## Cevaplar

```python
# 1
def kare(x):
    return x * x
print(kare(7))

# 2
def fahrenheit(c):
    return c * 9 / 5 + 32
for c in [0, 25, 100]:
    print(c, "°C =", fahrenheit(c), "°F")

# 3
def cift_say(liste):
    sayac = 0
    for n in liste:
        if n % 2 == 0:
            sayac += 1
    return sayac
print(cift_say([1, 2, 3, 4, 10, 15]))

# 4
def indirim(fiyat, oran=10):
    return fiyat * (100 - oran) / 100
print(indirim(200))
print(indirim(200, 25))
```
