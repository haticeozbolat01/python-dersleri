## İndeks: sıradaki elemana ulaşmak

Python saymaya 0'dan başlar. `-1` son eleman.

```python
notlar = [70, 85, 40, 90]
#  indeks: 0   1   2   3

print(notlar[0])    # 70
print(notlar[2])    # 40
print(notlar[-1])   # 90
print(notlar[-2])   # 40
```

## Dilimleme

Başlangıç dahil, bitiş dahil değil.

```python
alisveris = ["süt", "ekmek", "yumurta", "peynir", "çay"]

print(alisveris[1:3])
print(alisveris[:2])
print(alisveris[2:])
print(alisveris[::-1])
```

## Eklemek ve silmek

```python
liste = ["süt", "ekmek"]

liste.append("çay")
print(liste)

liste.insert(0, "yumurta")
print(liste)

liste.remove("ekmek")
print(liste)

son = liste.pop()
print(son)
print(liste)
```

## Sıralamak, en büyük, en küçük, toplam

```python
notlar = [70, 85, 40, 90]

print(sorted(notlar))   # yeni, sıralı liste
print(notlar)           # asıl liste değişmedi

notlar.sort()
print(notlar)

notlar.sort(reverse=True)
print(notlar)
```

```python
print(min(notlar))
print(max(notlar))
print(sum(notlar))
print(sum(notlar) / len(notlar))
```

## Kopyalama tuzağı

```python
a = [1, 2]
b = a
b.append(3)
print(a)         # a da değişti!

c = a.copy()
c.append(4)
print(a, c)
```

## Mini program: "bitti" diyene kadar not

Çalıştırınca not girmeni isteyecek. Bitirmek için `bitti` yaz.

```python
notlar = []

while True:
    giris = input("Not gir (bitirmek için 'bitti' yaz): ")
    if giris == "bitti":
        break
    notlar.append(float(giris))

print(f"{len(notlar)} not girdin.")
print(f"En yüksek: {max(notlar)}, en düşük: {min(notlar)}")
print(f"Ortalama: {sum(notlar) / len(notlar):.1f}")

gecenler = []
for n in notlar:
    if n >= 50:
        gecenler.append(n)
print("Geçen notlar:", gecenler)
```

## Cevaplar

```python
# 1
sayilar = [3, 8, 1, 9, 4]
print(sum(sayilar) / len(sayilar))
print(max(sayilar) - min(sayilar))

# 2
renkler = ["kırmızı", "mavi", "yeşil"]
renkler.append("sarı")
renkler.remove("mavi")
print(renkler, len(renkler))

# 3
harfler = ["p", "y", "t", "h", "o", "n"]
print(harfler[:3])
print(harfler[-3:])

# 4
tekrar = [1, 2, 2, 3, 1, 4]
tekil = []
for s in tekrar:
    if s not in tekil:
        tekil.append(s)
print(tekil)
```
