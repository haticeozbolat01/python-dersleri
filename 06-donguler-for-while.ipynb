## for: her biri için

```python
notlar = [70, 85, 90]

for n in notlar:
    print(n)
```

```python
for harf in "Python":
    print(harf, end=" ")
```

## range: sayı üretmek

`range` son sayıyı dahil etmez.

```python
for i in range(5):
    print(i, end=" ")
print()
print(list(range(1, 6)))
print(list(range(0, 10, 2)))
```

```python
for i in range(1, 11):
    print(f"7 x {i} = {7 * i}")
```

## Biriktirmek: döngüde toplam almak

```python
notlar = [70, 85, 90]
toplam = 0                     # kumbara boş başlar

for n in notlar:
    toplam += n                # toplam = toplam + n demek

print("Toplam:", toplam)
print("Ortalama:", toplam / len(notlar))
```

## while: koşul doğru olduğu sürece

```python
sayac = 3

while sayac > 0:
    print(sayac)
    sayac -= 1                 # her turda bir azalt

print("Kalkış!")
```

## break ve continue

```python
notlar = [70, 85, 40, 90]

for n in notlar:
    if n < 50:
        print("Kalan var:", n)
        break

for n in notlar:
    if n < 50:
        continue
    print("Geçti:", n)
```

## Mini program: istediğin kadar not

Çalıştırınca önce kaç not gireceğini soracak.

```python
adet = int(input("Kaç not gireceksin? "))
toplam = 0

for i in range(1, adet + 1):
    not_ = float(input(f"{i}. not: "))
    toplam += not_

ortalama = toplam / adet
print(f"Ortalaman: {ortalama:.1f}")

if ortalama >= 50:
    print("Geçtin!")
else:
    print("Kaldın.")
```

2. `sayi = int(input("Bir sayı: "))`, sonra `for i in range(1, 11): print(f"{sayi} x {i} = {sayi * i}")`.
3. `for i in range(1, 21):` içinde `if i % 2 != 0: continue` ve `print(i)`. Ya da `for i in range(2, 21, 2): print(i)`.
4. Aşağıdaki hücrede.

```python
gizli = 7
deneme = 0

while True:
    tahmin = int(input("Tahminin: "))
    deneme += 1
    if tahmin == gizli:
        print(f"Bildin! {deneme} denemede.")
        break
    elif tahmin < gizli:
        print("Daha büyük")
    else:
        print("Daha küçük")
```
