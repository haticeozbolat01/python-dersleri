## Geçen dersin sınıfı

Ders 17'deki `Gorev` sınıfıyla başlıyoruz.

```python
class Gorev:
    def __init__(self, baslik):
        self.baslik = baslik
        self.bitti = False

    def tamamla(self):
        self.bitti = True

    def __str__(self):
        isaret = "[x]" if self.bitti else "[ ]"
        return f"{isaret} {self.baslik}"
```

## Kalıtım: var olanın üzerine kurmak

Şimdi son tarihi olan bir görev istiyoruz. `Gorev`'in her şeyini yeniden yazmak yerine ondan **miras alıyoruz**: Yeni sınıf, eski sınıfın bütün özelliklerini ve metotlarını hazır alıyor, biz sadece farklı olanı ekliyoruz.

```python
from datetime import date


class SonTarihliGorev(Gorev):
    def __init__(self, baslik, son_tarih):
        super().__init__(baslik)
        self.son_tarih = son_tarih

    def gecikti_mi(self, bugun):
        return not self.bitti and bugun > self.son_tarih

    def __str__(self):
        temel = super().__str__()
        return f"{temel} (son tarih: {self.son_tarih})"


odev = SonTarihliGorev("Python ödevi", date(2026, 11, 20))
print(odev)
odev.tamamla()
print(odev)
```

Satır satır:

- `class SonTarihliGorev(Gorev):` Parantezin içi "bu sınıf Gorev'den miras alıyor" demek. `Gorev` **üst sınıf**, `SonTarihliGorev` **alt sınıf**.
- `super().__init__(baslik)` Üst sınıfın `__init__`'ini çalıştırıyor. Böylece `baslik` ve `bitti` özelliklerini yeniden yazmamıza gerek kalmadı.
- `tamamla()` metodunu hiç yazmadık ama `odev.tamamla()` çalıştı: Üst sınıftan geldi.
- `__str__` metodunu yeniden yazdık. Buna **ezme** (override) deniyor. İçinde `super().__str__()` ile önce üst sınıfın yazdırdığını alıp sonuna son tarihi ekledik.

## Gecikme kontrolü

```python
kontrol_gunu = date(2026, 11, 22)

rapor = SonTarihliGorev("Rapor yaz", date(2026, 11, 18))
sunum = SonTarihliGorev("Sunum hazırla", date(2026, 11, 25))

print(rapor.baslik, "gecikti mi?", rapor.gecikti_mi(kontrol_gunu))
print(sunum.baslik, "gecikti mi?", sunum.gecikti_mi(kontrol_gunu))
```

`gecikti_mi` metoduna bugünün tarihini dışarıdan veriyoruz. Böylece sonuç, kodu hangi gün çalıştırdığına bağlı olmuyor ve test etmesi kolay oluyor. Gerçek kullanımda `date.today()` verebilirsin.

## Farklı nesneler, aynı çağrı

Normal görevleri ve son tarihli görevleri aynı listede tutabilir, hepsini aynı şekilde yazdırabiliriz. Her biri kendi `__str__` metodunu kullanır.

```python
gorevler = [
    Gorev("Kitap oku"),
    SonTarihliGorev("Vergi beyannamesi", date(2026, 11, 30)),
    Gorev("Spor yap"),
]
gorevler[0].tamamla()

for g in gorevler:
    print(g)
```

Döngü, hangi nesnenin hangi sınıftan olduğunu bilmiyor; sadece `print(g)` diyor. Doğru `__str__` kendiliğinden seçiliyor. Buna **çok biçimlilik** (polymorphism) deniyor.

## isinstance: Bu nesne hangi sınıftan?

```python
for g in gorevler:
    print(g.baslik, "->", isinstance(g, Gorev), isinstance(g, SonTarihliGorev))
```

Dikkat: Son tarihli görev aynı zamanda bir `Gorev`. Alt sınıfın her nesnesi, üst sınıfın da bir nesnesidir. Kurabiye benzetmesiyle: Çikolatalı kurabiye de bir kurabiye.

## Alıştırmalar

1. Ders 17'deki `Kitap` sınıfını yaz (`baslik`, `yazar`, `sayfa`). Ondan miras alan bir `EKitap` sınıfı yaz; fazladan `dosya_boyutu_mb` özelliği olsun.
2. `EKitap`'ın `__str__` metodunu ezip sonuna dosya boyutunu ekle. `super().__str__()` kullan.
3. Bir listede iki `Kitap` ve bir `EKitap` olsun; döngüyle hepsini yazdır ve kaç tanesinin e-kitap olduğunu `isinstance` ile say.

## Cevaplar

```python
class Kitap:
    def __init__(self, baslik, yazar, sayfa):
        self.baslik = baslik
        self.yazar = yazar
        self.sayfa = sayfa

    def __str__(self):
        return f"{self.baslik} - {self.yazar} ({self.sayfa} sayfa)"


class EKitap(Kitap):
    def __init__(self, baslik, yazar, sayfa, dosya_boyutu_mb):
        super().__init__(baslik, yazar, sayfa)
        self.dosya_boyutu_mb = dosya_boyutu_mb

    def __str__(self):
        return f"{super().__str__()} [{self.dosya_boyutu_mb} MB]"


raf = [
    Kitap("Örnek Roman", "Ad Soyad", 240),
    EKitap("Örnek Rehber", "Ad Soyad", 180, 3.2),
    Kitap("Örnek Deneme", "Ad Soyad", 120),
]
for k in raf:
    print(k)
print("E-kitap sayısı:", sum(isinstance(k, EKitap) for k in raf))
```
