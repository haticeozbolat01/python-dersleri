## Adım 1: Rastgele bir sayı tutmak

```python
import random

gizli = random.randint(1, 100)
print(gizli)
```

## Adım 2: Tek bir tahmin

```python
gizli = random.randint(1, 100)
tahmin = int(input("1 ile 100 arasında bir sayı tuttum. Tahminin: "))

if tahmin == gizli:
    print("Bildin!")
elif tahmin < gizli:
    print("Daha büyük bir sayı söyle.")
else:
    print("Daha küçük bir sayı söyle.")
```

## Adım 3: Bilene kadar sormak

```python
gizli = random.randint(1, 100)
deneme = 0

while True:
    tahmin = int(input("Tahminin: "))
    deneme += 1
    if tahmin == gizli:
        print(f"Bildin! {deneme} denemede buldun.")
        break
    elif tahmin < gizli:
        print("Daha büyük.")
    else:
        print("Daha küçük.")
```

## Adım 4: Hatalı girişe karşı korumak

```python
def sayi_iste(mesaj, en_az, en_cok):
    """Kullanıcıdan en_az ile en_cok arasında bir tam sayı alana kadar sorar."""
    while True:
        cevap = input(mesaj).strip()
        if not cevap.isdigit():
            print("Lütfen sadece rakam yaz.")
            continue
        sayi = int(cevap)
        if en_az <= sayi <= en_cok:
            return sayi
        print(f"Sayı {en_az} ile {en_cok} arasında olmalı.")

print("Aldığım sayı:", sayi_iste("Bir sayı yaz: ", 1, 100))
```

## Adım 5: Zorluk seviyesi ve deneme hakkı

```python
ZORLUKLAR = {
    "kolay": {"ust_sinir": 50, "hak": 10},
    "orta": {"ust_sinir": 100, "hak": 7},
    "zor": {"ust_sinir": 1000, "hak": 10},
}

def oyna(zorluk="orta"):
    """Bir el sayı tahmin oyunu oynatır. Kazanırsa deneme sayısını, kaybederse None döndürür."""
    ayar = ZORLUKLAR[zorluk]
    ust = ayar["ust_sinir"]
    gizli = random.randint(1, ust)
    print(f"1 ile {ust} arasında bir sayı tuttum. {ayar['hak']} hakkın var.")

    for deneme in range(1, ayar["hak"] + 1):
        tahmin = sayi_iste(f"{deneme}. tahmin: ", 1, ust)
        if tahmin == gizli:
            print(f"Bildin! {deneme} denemede buldun.")
            return deneme
        elif tahmin < gizli:
            print("Daha büyük.")
        else:
            print("Daha küçük.")

    print(f"Hakların bitti. Sayı {gizli} idi.")
    return None
```

## Adım 6: Ana program

```python
def ana_program():
    skorlar = []
    while True:
        zorluk = input("Zorluk seç (kolay / orta / zor): ").strip().lower()
        if zorluk not in ZORLUKLAR:
            print("Bu seviyeyi tanımıyorum, orta ile başlıyoruz.")
            zorluk = "orta"

        sonuc = oyna(zorluk)
        if sonuc is not None:
            skorlar.append(sonuc)

        if input("Tekrar oynamak ister misin? (e/h): ").strip().lower() != "e":
            break

    print(f"{len(skorlar)} oyun kazandın.")
    if skorlar:
        print(f"En iyi skorun: {min(skorlar)} deneme")

ana_program()
```

## Bonus: Oyunu bilgisayara oynatmak

Her seferinde aralığın tam ortasını söyleyen bir oyuncu. Bu yönteme **ikili arama** (binary search) deniyor.

```python
def bilgisayar_oyna(gizli, ust=100):
    alt, ust_sinir = 1, ust
    deneme = 0
    while True:
        tahmin = (alt + ust_sinir) // 2
        deneme += 1
        if tahmin == gizli:
            return deneme
        elif tahmin < gizli:
            alt = tahmin + 1
        else:
            ust_sinir = tahmin - 1

print(bilgisayar_oyna(37))
print(bilgisayar_oyna(100))

denemeler = [bilgisayar_oyna(s) for s in range(1, 101)]
print("En kötü durum:", max(denemeler), "deneme")
print("Ortalama:", sum(denemeler) / len(denemeler), "deneme")
```
