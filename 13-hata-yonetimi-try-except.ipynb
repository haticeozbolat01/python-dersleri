## try ve except

```python
try:
    sayi = int("abc")
    print("Bu satır çalışmaz.")
except ValueError:
    print("Bu bir sayı değil.")

print("Program devam ediyor.")
```

## Farklı hatalar, farklı cevaplar

```python
def bol(a, b):
    try:
        sonuc = int(a) / int(b)
    except ValueError:
        return "Lütfen sayı gir."
    except ZeroDivisionError:
        return "Sıfıra bölünemez."
    return sonuc

print(bol("10", "2"))
print(bol("on", "2"))
print(bol("10", "0"))
```

## Hata mesajını yakalamak: as

```python
notlar = {"Ayşe": 85, "Mert": 45}

try:
    print(notlar["Zeynep"])
except KeyError as hata:
    print("Bulunamayan anahtar:", hata)

try:
    print([1, 2, 3][10])
except IndexError as hata:
    print("Hata:", hata)
```

## else ve finally

```python
def oku(metin):
    try:
        sayi = int(metin)
    except ValueError:
        print(f"{metin!r} sayıya çevrilemedi.")
    else:
        print(f"Başarılı: {sayi}")
    finally:
        print("Kontrol bitti.\n")

oku("42")
oku("kırk iki")
```

## Kendi hatanı fırlatmak: raise

```python
def yas_kaydet(yas):
    if yas < 0:
        raise ValueError("Yaş negatif olamaz.")
    return f"Yaş kaydedildi: {yas}"

print(yas_kaydet(25))

try:
    yas_kaydet(-3)
except ValueError as hata:
    print("Hata:", hata)
```

## Kötü bir alışkanlık: her hatayı susturmak

```python
# Bunu yapma
def ortalama_kotu(notlar):
    try:
        return sum(notlar) / len(notlar)
    except:
        return 0

print(ortalama_kotu([]))          # 0: boş liste
print(ortalama_kotu("doksan"))    # 0: yanlış tip, ama sebebini hiç öğrenemedik

# Bunu yap
def ortalama(notlar):
    try:
        return sum(notlar) / len(notlar)
    except ZeroDivisionError:
        return None

print(ortalama([]))
```

## Mini program: güvenli not girişi

Ders 10'daki `sayi_iste` fonksiyonunu try ve except ile yeniden yazıyoruz. Bu sürüm eksi ve ondalıklı sayıları da doğru yakalıyor.

```python
def not_iste(mesaj):
    """0 ile 100 arasında bir not alana kadar sorar."""
    while True:
        cevap = input(mesaj)
        try:
            puan = float(cevap.replace(",", "."))
        except ValueError:
            print("Lütfen bir sayı yaz.")
            continue
        if 0 <= puan <= 100:
            return puan
        print("Not 0 ile 100 arasında olmalı.")

notlar = [not_iste(f"{i}. not: ") for i in range(1, 4)]
print("Ortalama:", round(sum(notlar) / len(notlar), 1))
```

## Cevaplar

```python
# 1
def guvenli_bol(a, b):
    try:
        return float(a) / float(b)
    except ValueError:
        return "Lütfen iki sayı gir."
    except ZeroDivisionError:
        return "Sıfıra bölünemez."
print(guvenli_bol("9", "3"), guvenli_bol("9", "x"), guvenli_bol("9", "0"))

# 2
sayilar, atlanan = [], 0
for m in ["10", "yirmi", "30", "", "40"]:
    try:
        sayilar.append(int(m))
    except ValueError:
        atlanan += 1
print(sayilar, "atlanan:", atlanan)

# 3
def bul(sozluk, anahtar):
    try:
        return sozluk[anahtar]
    except KeyError:
        return "Bulunamadı"
print(bul({"a": 1}, "a"), bul({"a": 1}, "b"))

# 4
def sifre_kontrol(sifre):
    if len(sifre) < 8:
        raise ValueError("Şifre en az 8 karakter olmalı.")
    return "Şifre uygun."
for s in ["kisa", "yeterince-uzun"]:
    try:
        print(sifre_kontrol(s))
    except ValueError as hata:
        print("Hata:", hata)
```
